# Plot the Available Ground Station Locations

```python
import pandas as pd
# load ground station data
filename = 'ground_station_data.csv'

# load the data
df = pd.read_csv(filename)

# display data
df.style
```

```python
# plot the locations on a map
import matplotlib.pyplot as plt
import pylupnt as pnt

# select antennas with dish diameter > 15m and with X band
df_moon = df[df['Dish size'] > 15]
df_moon = df_moon[df_moon['frequency_band'] == 'X']

# plot the locations of the antennas
lat = df_moon['latitude_deg'].values
lon = df_moon['longitude_deg'].values
names = df_moon['name'].values

# replace the longitude >180 with negative values
lon = [x if x < 180 else x - 360 for x in lon]

df_moon.style
```

```python
fig, ax = plt.subplots(figsize=(12, 6))
ax.set_xlim(-180, 180)
ax.set_ylim(-90, 90)
ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")
ax.set_title("Ground Station Locations witg Dish Diameter > 15m and X Band")

# load earth topo image
img = plt.imread(pnt.LUPNT_DATA_PATH + "/topo" + "/earth_surface.jpg")
# flip the image vertically and horizontally
img = img[::-1, ::-1]
ax.imshow(img, extent=[-180, 180, -90, 90], alpha=0.2)

figobj = [None] * len(names)

for i, txt in enumerate(names):
    # if text starts with DSS
    if txt.startswith('DSS'):
        color = 'red'
        idx = 0
    if txt.startswith('LEGS'):
        color = 'blue'
        idx = 1
    if txt.startswith('IDSN'):
        color = 'green'
        idx = 2
    if txt.startswith('MDSS') or txt.startswith('USC'):
        color = 'orange'
        idx = 3
    if txt.startswith('GHY'):
        color = 'purple'
        idx = 4
    if txt.startswith('DSA'):
        color = 'black'
        idx = 5
    figobj[idx] = ax.scatter(lon[i], lat[i], color=color)
    ax.annotate(txt, (lon[i], lat[i]), color=color)

# add legend
ax.legend(figobj, ['DSS', 'LEGS', 'IDSN', 'MDSS/USC', 'GHY', 'DSA'], loc='best', fontsize='small')
plt.show()
```
